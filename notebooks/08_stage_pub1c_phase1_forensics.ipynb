# Pub-1C Phase-1 saved-result audit
This notebook reads saved production outputs only. It performs no simulation, selection, or training.

```python
from pathlib import Path
import json
import pandas as pd
ROOT = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / '.git').exists())
DATA = ROOT / 'outputs/stage_pub1c/data'
provenance = json.loads((DATA / 'provenance.json').read_text())
assert provenance['learned_training_steps'] == 0
assert provenance['primary_rho'] == 0.375
provenance['reuse']
```

```python
fractionality = pd.read_csv(DATA / 'fractionality_summary.csv')
energy = pd.read_csv(DATA / 'fractionality_energy_summary.csv')
fractionality.query("fractional_axis == 'joint' and epsilon in [0.0, 0.5] and modulation == '16QAM'")[[
    'epsilon', 'waveform', 'ber', 'total_bits'
]].sort_values(['epsilon', 'ber'])
```

```python
budget = pd.read_csv(DATA / 'budget_summary.csv')
retained = pd.read_csv(DATA / 'budget_energy_summary.csv')
assert set(budget['rho']) == set(retained['rho'])
budget.query("rho in [0.375, 1.0]")[[
    'modulation', 'rho', 'waveform', 'ber', 'total_bits'
]].sort_values(['modulation', 'rho', 'ber'])
```

```python
constellation15 = pd.read_csv(DATA / 'channel_type_constellation_summary_15db.csv')
constellation15_meta = json.loads((DATA / 'channel_type_constellation_channels_15db.json').read_text())
assert constellation15_meta['run_kind'] == 'production'
assert constellation15_meta['candidate_channels_per_regime'] == 256
assert constellation15_meta['p3_byte_identical']
assert (constellation15['support_nnz_per_row_min'] == 24).all()
assert (constellation15['support_nnz_per_row_max'] == 24).all()
constellation15[['regime_label', 'waveform', 'evm_rms', 'ber', 'ser']].sort_values(['regime_label', 'ber'])
```

```python
graphs = pd.read_csv(DATA / 'authoritative_threshold_graph_metrics.csv')
trained = json.loads((DATA / 'training_k_community_summaries.json').read_text())
display(graphs[['k', 'community_count', 'community_sizes', 'weighted_modularity', 'within_energy_fraction']])
{k: {'groups': v['group_count'], 'sizes': v['group_sizes']} for k, v in trained.items()}
```
