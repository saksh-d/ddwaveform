## Scientific classification

**Outcome C, qualified.** The simple receiver-budget law is not supported: no start follows 4→3→2, and the validation-selected sequence is 6→5→3. Partitions are initialization-sensitive, but the Haar runs are not communication-equivalent to the selected OFDM runs, so this stage does not establish multiple equivalent symmetry solutions. Matched K=16 training clearly repairs the frozen-waveform weakness; that communication specialization does not imply the proposed group-count law. STOP.
